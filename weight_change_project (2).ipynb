```python
#addressing and Handling Outliers
from scipy import stats
import numpy as np
zscore = np.abs(stats.zscore(df['Weight Change (lbs)']))

print(zscore)

keepNonOutliers=df[(zscore < 3)]

print('Before removing outliers', df.shape)
print("After removing outliers(Not Outliers): ", keepNonOutliers.shape)
print("-----------------------------------")
```

```python
# Identify and handle outliers using the IQR method

# Function to detect outliers based on IQR
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    return data[(data[column] < lower_bound) | (data[column] > upper_bound)]

# Numerical columns to check for outliers
numerical_cols = ['Daily Calories Consumed', 'Daily Caloric Surplus/Deficit', 'Current Weight (lbs)',
                 'BMR (Calories)', 'Weight Change (lbs)','Stress Level','Sleep Quality','Physical Activity Level']

# Detect outliers in each numerical column
outliers = {}
for col in numerical_cols:
    outliers[col] = detect_outliers_iqr(df, col)

# Print the number of outliers detected in each column
for col, outlier_data in outliers.items():
    print("Column: " + col + ", Outliers detected: " + str(len(outlier_data)))

# Handle outliers by capping them to the lower and upper bounds
def cap_outliers(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    data[column] = data[column].clip(lower=lower_bound, upper=upper_bound)

# Apply capping to all numerical columns
for col in numerical_cols:
    cap_outliers(df, col)

print("Outliers handled by capping to IQR bounds.")
```

```python
# Handle outliers by capping them to the IQR bounds
def handle_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1
    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR
    data[column] = data[column].clip(lower=lower_bound, upper=upper_bound)
    return data

# Handle outliers for all numerical columns
for col in numerical_cols:
    df = handle_outliers_iqr(df, col)

# Create box plots after handling outliers
plt.figure(figsize=(15, 10))
df.boxplot(column=numerical_cols)
plt.xticks(rotation=45)
plt.title('Box Plots After Outlier Handling')
plt.tight_layout()
plt.show()

print("\
Outliers have been handled by capping them to IQR bounds.")
print("Summary of the cleaned data:")
print(df[numerical_cols].describe())
```

# Formulating Hypothesises and Testing
 **Hypothesis 1**:
  -
- Null Hypothesis (H0)     :
   The magnitude of caloric surplus/deficit is not associated with the rate of weight change.
- Alternat Hypothesisiva H1:
  The magnitude of caloric surplus/deficit is associated with the rate of weight change.
  
  
   **Hypothesis 2:**
  -   Null hypothesis (H0):
      There is not a significant difference in weight change between different physical activity level

 - Alternative Hypotheis:
      There is a significant difference in weight change between different physical activity level

* For Hypothesis 2, we'll use ANOVA test since we're comparing weight changes across different activity levels.




**ANOVA Test:** The Analysis of Variance (ANOVA) test is used to determine if there are statistically significant differences between the means of three or more independent groups. In this case, we are comparing the weight change across different physical activity levels.

F-statistic: The F-statistic is a ratio that compares the variance between the groups to the variance within the groups. A higher F-statistic indicates a greater degree of difference between the group means relative to the variability within the groups.

P-value: The p-value indicates the probability of observing the data (or something more extreme) if the null hypothesis is true. A common threshold for significance is 0.05:

If the p-value is less than 0.05, we reject the null hypothesis, suggesting that there is a significant difference between the groups.
If the p-value is greater than 0.05, we fail to reject the null hypothesis, indicating that there is not enough evidence to suggest a significant difference across activity levels.

```python
# Testing Hypothesis 1:
from scipy.stats import pearsonr
import pandas as pd


columns_of_interest = ['Daily Caloric Surplus/Deficit', 'Weight Change (lbs)']


# Calculate Pearson correlation coefficient and p-value
correlation, p_value = pearsonr(df['Daily Caloric Surplus/Deficit'], df['Weight Change (lbs)'])

# Display results
print("\nHypothesis 1: The magnitude of caloric surplus/deficit is associated with the rate of weight change.")
print(f"Pearson Correlation Coefficient: {correlation:.2f}")
print(f"P-Value: {p_value:.4f}")

# Interpretation of results
if p_value < 0.05:
    print("\nConclusion: The null hypothesis is rejected. There is a significant association between caloric surplus/deficit and weight change.")
else:
    print("\nConclusion: The null hypothesis cannot be rejected. There is no significant association between caloric surplus/deficit and weight change.")
```

Interpretation of Results:

In this case, the Pearson correlation test resulted in a correlation coefficient of 0.06 and a p-value of 0.5463 Since the p-value (0.5463) is greater than 0.05, we fail to reject the null hypothesis for Hypothesis 1. This means that there is no statistically significant association between the magnitude of caloric surplus/deficit and the rate of weight change.

Hypothesis 1: Fail to reject the null hypothesis based on the results, indicating that: There is no significant association between caloric surplus/deficit and weight change.

```python
# Testing Hypothesis 2: ANOVA for weight change across physical activity levels
from scipy.stats import f_oneway

# Group weight change by physical activity levels
anova_groups = [df[df['Physical Activity Level'] == level]['Weight Change (lbs)'] for level in df['Physical Activity Level'].unique()]

# Perform ANOVA test
f_stat, p_value_anova = f_oneway(*anova_groups)

print("\
Hypothesis 2 Results:")
print("F-statistic: " + str(round(f_stat, 4)))
print("P-value: " + str(round(p_value_anova, 4)))
```

* Interpretation of Results:

In this case, the ANOVA test resulted in an F-statistic of 0.3911and a p-value of 0.7596 Since the p-value (0.7596) is greater than 0.05, we fail to reject the null hypothesis for Hypothesis 2. This means that there is no statistically significant difference in weight change across different physical activity levels.


Hypothesis 2: fail to reject the null hypothesis based on the results, indicating that:
There is no significant difference in weight change across different physical activity levels.

# Model implemenatation and analysis:
This code prepares the data for a machine learning model by:
* 1-
Separating the predictive features (X) from the target variabe ( weight_changey
* 2- .
Splitting the data into training and testing sets to train and evaluate the modeets.

```python
# variable 'X' holds all the feature columns & variable 'y' holds the target column.
X = df.drop(labels='Weight Change (lbs)', axis=1)  #axis=1 means we drop data by column.

#dependent variable / response / target variable.
y = df['Weight Change (lbs)']

# Split data into training (80%) and testing (20%) sets
train_X, test_X, train_y, test_y = train_test_split(X,y,test_size=0.20,random_state=999)
print(train_X.shape)
print(test_X.shape)
print(train_y.shape)
print(test_y.shape)
```

### Training the model

```python
# 1- Create instance of LinearRegression
lm = LinearRegression()
print(lm)
print(type(lm))

# 2- Train the model
lm.fit(train_X, train_y) # fit the model into the given dataset (finding coefficients and intercept)
```

```python
# Predict the progression of weight change based on test data
predicted_y = lm.predict(test_X) # used the trained model to predict the output of new data points
result = pd.DataFrame({'Actual': test_y, 'Predict' : predicted_y})

result
```

### Evaluating the model

```python
print("The R-Square score of the model is: " , np.round(metrics.r2_score(test_y,predicted_y),2))
print()
print('coefficient:', lm.coef_)
print()
print('intercept:', lm.intercept_)
```

### **Random Forest Model**



1. **Import RandomForestRegressor:**
 We import the RandomForestRegressor class from sklearn.ensemble.

2.  **Create and Train the Model**:
 We create an instance of the RandomForestRegressor and train it using the training data (train_X, train_y)

3. **Predict and Evaluate:**
 We use the trained model to predict on the test data (test_X) and then evaluate the performance using the same regression metrics (MSE, MAE, R-squared).


The Random Forest Regressor was chosen because of its adaptability. In comparison to Linear Regression, Random Forest  perform better since it can identify non-linear relationships in the data. It is also less likely to overfit. Additionally, it provides the significance of several features in predicting the desired variable.

```python
# Import necessary libraries
from sklearn.ensemble import RandomForestRegressor
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, mean_absolute_error, r2_score

# Assuming features in 'X' and target variable in 'y'

# Split data into training (80%) and testing (20%) sets
train_X, test_X, train_y, test_y = train_test_split(X, y, test_size=0.20, random_state=999)

# Create a Random Forest Regressor instance
rf_regressor = RandomForestRegressor(n_estimators=100, random_state=999)

# Train the model
rf_regressor.fit(train_X, train_y)

# Predict on the test set
predicted_y_rf = rf_regressor.predict(test_X)

# Evaluate the model
mse_rf = mean_squared_error(test_y, predicted_y_rf)
mae_rf = mean_absolute_error(test_y, predicted_y_rf)
r2_rf = r2_score(test_y, predicted_y_rf)

# Print the evaluation metrics
print("Random Forest Regressor Performance:")
print(f"Mean Squared Error (MSE): {mse_rf:.2f}")
print(f"Mean Absolute Error (MAE): {mae_rf:.2f}")
print(f"R-squared (R2): {r2_rf:.2f}")
```

```python
# Predict using the Random Forest model
predicted_y_rf = rf_regressor.predict(test_X)

# Create results dataframe
result = pd.DataFrame({'Actual': test_y, 'Predicted': predicted_y_rf})
print(result)
```
